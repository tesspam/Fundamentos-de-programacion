Ejercicio extra 1

```python

def suma_matriz(matriz):
    suma = 0
    for fila in matriz:
        for elemento in fila:
            suma = suma + elemento
    return suma

def imprimir_matriz(matriz):
    for fila in matriz:
        for elemento in fila:
            print(elemento, end=" ")
        print()

matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

imprimir_matriz(matriz)

resultado = suma_matriz(matriz)

print("La suma de todos los elementos de la matriz es:", resultado)


```

Ejercicio extra 2

```python

def suma_fila(fila):
    suma = 0
    for elemento in fila:
        suma = suma + elemento
    return suma


def mostrar_tabla(matriz):
    for fila in matriz:
        for elemento in fila:
            print(elemento, end=" ")
        print()


matriz = [[3, 1, 4], [1, 5, 9], [2, 6, 5]]

mostrar_tabla(matriz)

for i in range(len(matriz)):
    resultado = suma_fila(matriz[i])
    print("Suma de la fila", i, ":", resultado)
```

Ejercicio extra 3

```python


def multiplicar_columna(matriz, columna):
    producto = 1

    for fila in matriz:
        numero = fila[columna]

        resultado = 0
        for i in range(numero):
            resultado = resultado + producto

        producto = resultado

    return producto


matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9], [2, 4, 6]]

columna = int(input("Numero de columna: "))

resultado = multiplicar_columna(matriz, columna)

print("El producto de la columna", columna, "es:", resultado)
```

Ejercicio extra 4

```python
def generar_tabla(tamano):
    tabla = []

    for fila in range(1, tamano + 1):
        renglon = []

        for columna in range(1, tamano + 1):
            resultado = 0

            for i in range(columna):
                resultado = resultado + fila

            renglon.append(resultado)

        tabla.append(renglon)

    return tabla


def imprimir_tabla(tabla):
    for fila in tabla:
        for elemento in fila:
            print(elemento, end="\t")
        print()


tamano = int(input("Tamano de la tabla: "))

if tamano < 2 or tamano > 5:
    print("Error: el tamaño debe estar entre 2 y 5.")
else:
    tabla = generar_tabla(tamano)

    imprimir_tabla(tabla)

    renglon = int(input("Renglon (factor): "))
    columna = int(input("Columna (factor): "))

    resultado = tabla[renglon - 1][columna - 1]

    print("El producto de", renglon, "x", columna, "es:", resultado)
```
