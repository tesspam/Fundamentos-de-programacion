Ejercicio  extra 1


```python

calificaciones = (7.5, 9.0, 8.0, 6.5, 10.0)

print("Tercera calificación:", calificaciones[2])

nueva1 = float(input("Nueva calificación 1: "))
nueva2 = float(input("Nueva calificación 2: "))

nueva_tupla = calificaciones + (nueva1, nueva2)

print("Nueva tupla:", nueva_tupla)

lista_ordenada = list(nueva_tupla)
lista_ordenada.sort(reverse=True)

print("Lista ordenada (mayor a menor):", lista_ordenada)

def calcular_suma(calificaciones):
    suma = 0
    for calificacion in calificaciones:
        suma += calificacion
    return suma

suma_total = calcular_suma(nueva_tupla)
print("Suma total:", suma_total)
```

Ejercicio extra 2

```python
agenda = {
    "Ana": "555-0101",
    "Luis": "555-0102",
    "Mía": "555-0103"
}
nombre= input("Ingrese el nombre de la persona que desea buscar:")
telefono= input ("Ingrese el número de teléfono del nuevo contacto:")

agenda[nombre]= telefono
print ("Contactos registrados: ", agenda)

def buscar_contacto(nombre):
    return agenda [nombre]

nombre_buscar= input ("Ingrese el nombre de la persona que desea buscar:")
telefono_encontrado=buscar_contacto(nombre_buscar)
print("El número de teléfono de", nombre_buscar, "es", telefono_encontrado)
```

```python


def contar_palabras(mensaje):
    palabras = mensaje.split()
    return len(palabras)
mensaje = input("Ingrese un mensaje: ")

print("Longitud del mensaje:", len(mensaje), "caracteres")
print ("En mayúsculas:" , mensaje.upper ())

texto_reemplazado = mensaje.replace("Python", "programación")
print ("Texto reemplazado :", texto_reemplazado)

print("Número de palabras en el mensaje:", contar_palabras(mensaje))





```
