Ejercicio 13

```python
def sumar_lista(lista):
    total=0
    for elemento in lista:
        total += elemento
    return total 
elemento = [10,20,30,40]
resultado = sumar_lista(elemento)
print(f"La suma de los elementos es: {resultado}")
```

Ejercico 14

```python
def obtener_maximo(lista):
    maximo = lista[0]
    for num in lista:
        if num in lista:
            if num > maximo:
                maximo = num 
    return maximo 

elementos = [7,3,9,2]
resultado = obtener_maximo (elementos)
print(f" El elemnto máximo es: {resultado}")

```

Ejercicio 18

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

numero = 5
resultado = factorial(numero)
print(f"El factorial de {numero} es: {resultado}")
```
