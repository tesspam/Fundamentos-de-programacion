```python
edad = int(input("Edad: "))

if edad >= 18:
    print("Eres mayor de edad")
else:
    print("Eres menor de edad")
```

```python
numero = int(input("Número: "))

if numero % 2 == 0:
    print(f"{numero} es un número par")
else:
    print(f"{numero} es un número impar")
```

```python
calificacion = float(input("Calificación: "))

if calificacion >= 6:
    print("Aprobado")
else:
    print("Reprobado")
```

```python
edad = int(input("Edad: "))

if edad < 3:
    print("Bebé")
elif edad <= 12:
    print("Niño")
elif edad <= 17:
    print("Adolescente")
else:
    print("Adulto")
```

```python
tipo = input("Tipo de visitante: ")

precio = 45

if tipo == "adulto_mayor":
    descuento = 0.12
elif tipo == "profesor":
    descuento = 0.10
elif tipo == "estudiante":
    descuento = 0.10
else:
    descuento = 0

precio_final = precio - (precio * descuento)

print(f"Precio final: ${precio_final:.2f}")
```

```python
edad = int(input("Edad: "))
credencial = input("¿Tiene credencial vigente? ")

if edad >= 18 and credencial == "si":
    print("Acceso permitido")
else:
    print("Acceso denegado")
```

```python
numero1 = float(input("Primer número: "))
numero2 = float(input("Segundo número: "))

if numero1 < 0 or numero2 < 0:
    print("Hay un número negativo")
else:
    print("No hay números negativos")
```

```python
numero = int(input("Número: "))

if numero > 0:
    if numero % 2 == 0:
        print("El número es positivo y par")
    else:
        print("El número es positivo pero impar")
else:
    print("El número no es positivo")
```

```python
respuesta = input("¿Desea continuar? ")

if not respuesta == "si":
    print("Proceso finalizado")
else:
    print("El proceso continúa")
```

```python
contador = 1

while contador <= 5:
    print(contador)
    contador += 1
```

```python
n = int(input("Ingrese N: "))

contador = 1
suma = 0

while contador <= n:
    suma += contador
    contador += 1

print(f"La suma del 1 al {n} es: {suma}")
```

```python
numero = int(input("Número: "))

contador = 1

while contador <= 10:
    resultado = numero * contador
    print(f"{numero} x {contador} = {resultado}")
    contador += 1
```
